```python
#W11A3
def is_happy_num(arr):
    try:
        sum = 0
        for i in arr:
            sum += int(i)**2
        if sum == 1:
            return 'YES'
        return is_happy_num(list(str(sum)))
    except ValueError:
        return('Loi nhap lieu')
    except RecursionError:
        return 'NO'

num = list(input())
print(is_happy_num(num))
    
    

```

```python
#W11A4
def tinh_tong_hai_so():
    try:
        a = float(input("Nhập số thứ nhất: "))
        b = float(input("Nhập số thứ hai: "))
    except ValueError:
        print("Lỗi: Bạn phải nhập số hợp lệ.")
    else:
        print("Tổng hai số là:", a + b)
    finally:
        print("Kết thúc chương trình.")

tinh_tong_hai_so()
```
