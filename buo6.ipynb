```python
#6A12
from collections import Counter
def ham(list1:list):
    dsmoi = []
    dict1 = Counter(list1)
    max1 = max(dict1.values())
    for keys,values in dict1.items():
        if values == max1:
            dsmoi.append(keys)
    return min(dsmoi)


list1 = list(map(int, input().split()))
print(ham(list1))
```
