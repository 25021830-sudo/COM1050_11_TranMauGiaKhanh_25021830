```python
#W10A4
def valid_transformation(a,b):
    b_index = 0
    a_index = 0
    while a_index <= len(a) and a_index <= len(b):
        if a.lower()[a_index] == b.lower()[b_index]:
            b_index += 1
        a_index += 1
    if b_index == len(b):
        return "YES"
    return "NO"
n = int(input())
ds = []
for i in range(n):
    a = input()
    b = input()
    ds.append(a)
    ds.append(b)
for i in range(0,n*2,2):
    print(valid_transformation(ds[i],ds[i+1]))
```

```python
#W10A5
def valid_in(mot_list,hai_list):
    for i in range(0,len(hai_list) - len(mot_list),1):
        if mot_list == hai_list[i:i+len(mot_list)]:
            return "YES"
    return "NO"
mot_list = list(map(int,input().split()))
hai_list = list(map(int,input().split()))
print(valid_in(mot_list,hai_list))
```

```python
#W10A6
def next_permutation(nums):
    pivot_idx = -1
    for i in range(len(nums)-2,-1,-1):
        if nums[i] < nums[i + 1]:
            pivot_idx = i
            break
    if pivot_idx == -1:
        nums.reverse()
        return nums
    for j in range(len(nums)-1,pivot_idx,-1):
        if nums[j] > nums[i]:
            succesor_idx = j
            break
    nums[pivot_idx],nums[succesor_idx] = nums[succesor_idx],nums[pivot_idx]
    # Reverse the complement
    nums[pivot_idx+1:] = sorted(nums[pivot_idx+1:])
    return nums
nums = list(map(int,input().split()))
print(next_permutation(nums))
```

```python
#W10A7
def canPlaceFlowers(flowerbed,k):

    check_pos = 0
    while check_pos < len(flowerbed):
        if check_pos == 0:
            if flowerbed[check_pos] == 0 and flowerbed[check_pos] == 0:
                k -= 1
                flowerbed[check_pos] = 1
                check_pos += 2
            else:
                check_pos += 2
        elif check_pos == len(flowerbed) - 1:
            if flowerbed[check_pos] == 0 and flowerbed[check_pos - 1] == 0:
                k -= 1
            check_pos += 1
        else:
            if flowerbed[check_pos] == 0:
                if flowerbed[check_pos - 1] != 1 and flowerbed[check_pos + 1] != 1:
                    k -= 1
                    flowerbed[check_pos] = 1
                    check_pos += 2
            elif flowerbed[check_pos] == 1:
                check_pos += 2


    if k == 0:
        return True
    return False
arr = list(map(int,input().split()))
k = int(input())
print(canPlaceFlowers(arr,k))
```

```python
#W10A8
def is_path_crossing(moves):
    # Vị trí ban đầu
    x, y = 0, 0
    # Set lưu trữ các tọa độ đã đi qua, khởi tạo với (0, 0)
    visited = {(0, 0)}

    for move in moves:
        if move == 'U': y += 1
        elif move == 'D': y -= 1
        elif move == 'R': x += 1
        elif move == 'L': x -= 1

        # Nếu tọa độ mới đã nằm trong set, nghĩa là đã lặp lại
        if (x, y) in visited:
            return True
        visited.add((x, y))

    return False


commands = input()
print(is_path_crossing(commands))
```

```python
#W10A9
def Median(nums):
    a, b, c, d, e = nums


    if a > b: a, b = b, a
    if c > d: c, d = d, c

    if a < c:

        a = e
        if a > b: a, b = b, a
    else:

        c = e
        if c > d: c, d = d, c


    if b < d:

        if a < c:
            return c if c < b else b
        else:
            return a if a < d else d
    else:
        if c < a:
            return a if a < d else d
        else:
            return c if c < b else b

nums = [0,1,2,3,4]
print(Median(nums))
```

```python
#W10A10 CHUA XONG
import bisect

def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    radius = 0

    for house in houses:

        idx = bisect.bisect_left(heaters, house)


        dist_right = heaters[idx] - house if idx < len(heaters) else float('inf')

        # Khoảng cách đến lò sưởi bên trái (nếu có)
        dist_left = house - heaters[idx-1] if idx > 0 else float('inf')

        # Khoảng cách nhỏ nhất cho ngôi nhà này
        current_min_dist = min(dist_left, dist_right)

        # r là giá trị lớn nhất trong các khoảng cách nhỏ nhất
        radius = max(radius, current_min_dist)

    return radius
houses = list(map(int,input().split()))
heaters = list(map(int,input().split()))
print(findRadius(houses,heaters))
```
