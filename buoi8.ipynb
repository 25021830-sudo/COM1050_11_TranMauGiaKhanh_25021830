```python
#W8A7
class Fraction:
    def __init__(self,a,b,op,c,d):
        self.a = int(a)
        self.b = int(b)
        self.c = int(c)
        self.d = int(d)
        self.op = op
        if self.a > 0:
            for i in range(int(min(self.a,self.b)**1/2),1,-1):
                if self.a% i == 0 and self.b % i == 0:
                    self.a //= i
                    self.b //= i
                    break
        elif self.a < 0:
            for i in range(self.a,-1,1):
                if self.a% i == 0 and self.b % i == 0:
                    self.a //= abs(i)
                    self.b //= abs(i)
                    break
        if self.c > 0:
            for i in range(int(min(self.c,self.d)**1/2),1,-1):
                if self.c% i == 0 and self.d % i == 0:
                    self.c //= i
                    self.d //= i
                    break
        elif self.c < 0:
            for i in range(self.c,-1,1):
                if self.c% i == 0 and self.d % i == 0:
                    self.c //= abs(i)
                    self.d //= abs(i)
                    break

    def __str__(self,tu_so,mau_so):
        if tu_so > 0:
            for i in range(int(min(tu_so,mau_so)),1,-1):
                if int(tu_so) % i == 0 and int(mau_so) % i == 0:
                    tu_so //= i
                    mau_so//= i
                    break
        if tu_so < 0:
            for i in range(-tu_so,1,-1):
                if abs(tu_so) %i == 0 and abs(mau_so) % i == 0:
                    tu_so //= i
                    mau_so //= i
                    break
        return f"{tu_so}/{mau_so}"


    def thuc_hien_tinh_toan(self):
        if self.op == "*":
            tu_so = self.a*self.c
            mau_so = self.b*self.d
        if self.op == "/":
            tu_so = self.a*self.d
            mau_so = self.b*self.c
        if self.op == "+":
            tu_so = self.a*self.d + self.c*self.b
            mau_so = self.b*self.d
        if self.op == "-":
            tu_so = self.a*self.d - self.c*self.b
            mau_so = self.b*self.d
        return self.__str__(tu_so,mau_so)
a,b,op,c,d = map(str,input().split(" "))

kk = Fraction(a,b,op,c,d)
kk.thuc_hien_tinh_toan()
```

```python
#W8A8
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = float(balance)
    def deposit(self,so):
        self.balance += float(so)
    def withdraw(self,so):
        if so <= self.balance:
            self.balance -= float(so)
    def __str__(self):
        return f"{self.balance:.2f}"
owner,balance = map(str,input().split())
kk = BankAccount(owner,balance)
so_lan_nhap = int(input())
for i in range(so_lan_nhap):
    lenh,so = map(str,input().split())
    if lenh == "DEPOSIT":
        kk.deposit(so)
    if lenh == "WITHDRAW":
        kk.withdraw(float(so))
print(kk.__str__())


```

```python
#W8A9
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, subject, score):
        self.scores.append(float(score))

    def avg(self):
        if not self.scores:
            return 0.0
        return sum(self.scores) / len(self.scores)

    def rank(self):
        dtb = self.avg()
        if dtb >= 8: return "Excellent"
        if dtb >= 6.5: return "Good"
        if dtb >= 5: return "Average"
        return "Poor"


name = input()
n = int(input())
st = Student(name)
for _ in range(n):
    sub, sc = input().split()
    st.add_score(sub, sc)

# Output: name avg rank
print(f"{st.name} {st.avg():.2f} {st.rank()}")
```

```python
#W8A10
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

class Library:
    def __init__(self):
        self.books = []

    def add(self, book):
        self.books.append(book)

    def count_by_author(self, name):
        return sum(1 for b in self.books if b.author == name)

    def find_by_year(self, y):
        return sum(1 for b in self.books if b.year == y)


lib = Library()
n = int(input())
for _ in range(n):
    line = input().split(maxsplit=1)
    cmd = line[0]

    if cmd == "ADD":
        title, author, year = line[1].split(';')
        lib.add(Book(title, author, year))
    elif cmd == "COUNT":
        print(lib.count_by_author(line[1]))
    elif cmd == "COUNTYEAR":
        print(lib.find_by_year(line[1]))
```

```python
#W8A11
import math

class Vector:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

    # Phép trừ hai vectơ (để tìm vectơ đoạn thẳng)
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y, self.z - other.z)

    # Tích có hướng (Cross Product)
    def cross(self, other):
        return Vector(
            self.y * other.z - self.z * other.y,
            self.z * other.x - self.x * other.z,
            self.x * other.y - self.y * other.x
        )

    # Tích vô hướng (Dot Product)
    def dot(self, other):
        return self.x * other.x + self.y * other.y + self.z * other.z

    # Độ dài vectơ
    def length(self):
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)

def solve():
    try:
        # Nhập tọa độ 4 điểm
        points = []
        for _ in range(4):
            points.append(Vector(*map(float, input().split())))

        A, B, C, D = points

        # Vectơ pháp tuyến n1 của mp(ABC)
        ab = B - A
        bc = C - B
        n1 = ab.cross(bc)

        # Vectơ pháp tuyến n2 của mp(BCD)
        # Lưu ý: cd = D - C
        cd = D - C
        n2 = bc.cross(cd)

        # Tính cosin của góc giữa hai pháp tuyến
        # Công thức: cos(phi) = (n1 . n2) / (|n1| * |n2|)
        dot_val = n1.dot(n2)
        len_n1 = n1.length()
        len_n2 = n2.length()

        # Góc giữa hai mặt phẳng
        cos_phi = dot_val / (len_n1 * len_n2)

        # Đảm bảo giá trị nằm trong khoảng [-1, 1] để tránh lỗi toán học
        cos_phi = max(-1.0, min(1.0, cos_phi))

        phi_rad = math.acos(cos_phi)
        phi_deg = math.degrees(phi_rad)

        print(f"{phi_deg:.2f}")

    except EOFError:
        pass

solve()
```

```python
#W8A12
class ComplexNumber:
    def __init__(self, initRealPart, initImaginaryPart):
        self.realPart = float(initRealPart)
        self.imaginaryPart = float(initImaginaryPart)

    def multiply(self, other):
        new_real = self.realPart * other.realPart - self.imaginaryPart * other.imaginaryPart
        new_imag = self.realPart * other.imaginaryPart + self.imaginaryPart * other.realPart
        return ComplexNumber(new_real, new_imag)


r1, i1 = map(float, input().split())
r2, i2 = map(float, input().split())

c1 = ComplexNumber(r1, i1)
c2 = ComplexNumber(r2, i2)
result = c1.multiply(c2)


print(f"{result.realPart:.0f} {result.imaginaryPart:.0f}")
```

```python
#W8A13
import math
class Triangle:
    def __init__(self, edge1, edge2, edge3):
        self.e1 = edge1
        self.e2 = edge2
        self.e3 = edge3

    def getArea(self):
        a, b, c = self.e1, self.e2, self.e3
        # Kiểm tra điều kiện tam giác
        if a + b > c and a + c > b and b + c > a:
            p = (a + b + c) / 2
            area = math.sqrt(p * (p - a) * (p - b) * (p - c))
            return area
        return -1

# Input
try:
    a, b, c = map(int, input().split())
    tri = Triangle(a, b, c)
    result = tri.getArea()

    if result == -1:
        print("invalid")
    else:
        # Nếu diện tích là số nguyên (như 6.0), in số nguyên, nếu không in 2 chữ số
        if result == int(result):
            print(int(result))
        else:
            print(f"{result:.2f}")
except:
    print("invalid")
```
